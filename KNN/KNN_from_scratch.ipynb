```python
import numpy as np
import pandas as pd
```

```python
from sklearn.datasets import make_classification,make_regression
```

```python
x_class,y_class = make_classification(n_samples = 200, n_features = 4, n_informative = 2,n_classes = 2,n_clusters_per_class = 2 )
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x_class,y_class,test_size=0.2,random_state=30)
```

```python
from sklearn.neighbors import KNeighborsClassifier
```

```python
knn = KNeighborsClassifier(n_neighbors=5)
```

```python
knn.fit(x_train,y_train)
```

```python
y_pred = knn.predict(x_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
print(x_test.shape)
print(x_train.shape)
```

```python

from  collections import Counter
class apnaKNN():
    def __init__(self,k):
        self.n_neighbors = k
        self.x_train = None
        self.y_train= None
        
    def fit(self,x_train,y_train):
        self.x_train = x_train
        self.y_train= y_train

    def predict(self,x_test):
        y_pred  = []
        for i in x_test:
            distances = []
            for j in x_train:
                distances.append(self.calculate_dist(i,j))
            n_neighbors= sorted(list(enumerate(distances)),key=lambda x:x[1])[0:self.n_neighbors]
            label = self.majority_count(n_neighbors)
            y_pred.append(label)
        return np.array(y_pred)

    
    def calculate_dist(self,point_a,point_b):
        return  np.linalg.norm(point_a - point_b)
        
    def majority_count(self,neighbors):
        votes = []
        for i in neighbors:
            votes.append(self.y_train[i[0]])
        votes = Counter(votes)
        return votes.most_common()[0][0]
      
        
   
        
       
        
```
