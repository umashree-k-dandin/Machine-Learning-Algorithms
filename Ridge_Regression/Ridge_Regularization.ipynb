This notebook compares Linear Regression and Ridge Regression on the diabetes dataset (442 samples, 10 features) to demonstrate the effect of L2 regularization. After splitting data into 80-20 train-test sets, Linear Regression achieved R² = 0.5188 with RMSE = 48.73, while Ridge Regression with alpha=0.02 achieved R² = 0.5233 with RMSE = 48.50, showing slight improvement. The Ridge model adds a penalty term α * Σ(θ²) to the loss function, shrinking coefficients to reduce overfitting and handle multicollinearity. 

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
#linear regresssion and the Ridge regression applied on the  A multiple-feature regression dataset

from sklearn.datasets import load_diabetes

data = load_diabetes()
```

```python
print(data.DESCR)
```

```python
x=data.data
y =data.target
print(x.shape,y.shape)
#plt.scatter(x,y)
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test =train_test_split(x,y,test_size=0.2,random_state=45)
```

```python
from sklearn.linear_model import LinearRegression
lr=LinearRegression()
```

```python
lr.fit(x_train,y_train)
```

```python
y_pred = lr.predict(x_test)
```

```python
from sklearn.metrics import r2_score,mean_squared_error
print("R2 score ( linear Regression)",r2_score(y_test,y_pred))
print("Mean squared error (using Linear Regression)",np.sqrt(mean_squared_error(y_test,y_pred)))
```

```python
from sklearn.linear_model import Ridge
R=Ridge(alpha=0.02)
```

```python
R.fit(x_train,y_train)
```

```python
y_pred1=R.predict(x_test)
```

```python
print("R2 Square after Ridge Regularization",r2_score(y_test,y_pred1))
print("Mean Squared Error after Ridge Regularization",np.sqrt(mean_squared_error(y_test,y_pred1)))
```
