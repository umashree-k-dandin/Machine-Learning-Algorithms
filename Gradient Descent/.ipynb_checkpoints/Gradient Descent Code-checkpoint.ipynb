```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
from sklearn.model_selection import cross_val_score
from sklearn.linear_model  import LinearRegression
```

```python
X,y = make_regression(n_samples=100,n_features=1,n_informative=1,noise=20,random_state=13)
```

```python
plt.scatter(X,y)
```

```python
from sklearn.model_selection import train_test_split
x_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
lr = LinearRegression()
```

```python
lr.fit(X,y)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
y_pred = lr.predict(X_test)
from sklearn.metrics import r2_score
R2_score=r2_score(y_test,y_pred)
print(R2_score)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X),color='red')
```

```python


class GDRegressor:
    def __init__(self,learning_rate,epochs):
        self.m = 100
        self.b = -120
        self.lr=learning_rate
        self.epochs=epochs

    def fit(self,X,y):
        # calculate b using the gradient descent
        for i in range(self.epochs):
          slope_b= -2 * np.sum(y - self.m*X.ravel() -  self.b)
          slope_m= -2 * np.sum((y - self.m*X.ravel() - self.b)* X.ravel())
            
          self.b = self.b- (self.lr*slope_b)    
          self.m = self.m- (self.lr*slope_m)
        print(self.b,self.m)
        
    def predict(self,X):
          return self.m * X + self.b
        
```

```python
gd = GDRegressor(0.001,100)
```

```python
gra=gd.fit(x_train,y_train)
```

```python
y_pred = gd.predict(X_test)
from sklearn.metrics import r2_score
R2_score=r2_score(y_test,y_pred)
print(R2_score)
```
