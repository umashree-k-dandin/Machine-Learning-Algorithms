```python
 mask = data[data['person_age'] > 60].index
data = data.drop(mask)
```

```python
cor1 = data.corr(numeric_only =True)
cor1['loan_status'].sort_values()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,OneHotEncoder,OrdinalEncoder
from sklearn.naive_bayes import GaussianNB,BernoulliNB,MultinomialNB
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score,classification_report,roc_auc_score
```

```python
x = data.iloc[:,:-1]
y  = data.iloc[:,-1]
```

```python
print(x.shape)
print(y.shape)
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
preprocessor = ColumnTransformer([
    ('scaler',StandardScaler(),['person_income','loan_amnt']),
    ('encoder',OneHotEncoder(),['person_gender']),
    ('Ord_encoder',OrdinalEncoder(),['person_education','person_home_ownership'])
])
```

```python
pipeline = Pipeline([
    ('preprocessor',preprocessor),
    ('classifier',GaussianNB()) #accuraccy score is 79
    #('bernoullinb',BernoulliNB()) # accuracy is 77.59
])
```

```python
pipeline.fit(x_train,y_train)
```

```python
y_pred = pipeline.predict(x_test)
```

```python
acc_score = accuracy_score(y_test,y_pred)
```

```python
print(acc_score)
```

```python
print(classification_report(y_test,y_pred))
```

```python
y_prob = pipeline.predict_proba(x_test)[:,1]
```

```python
roc_auc_score(y_test,y_prob)
```
