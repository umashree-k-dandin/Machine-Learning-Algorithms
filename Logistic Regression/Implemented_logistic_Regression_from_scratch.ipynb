```python
from sklearn.datasets import make_classification
import numpy as np
```

```python
x,y = make_classification(n_samples=500,n_features=2,n_informative=1,n_redundant=0,n_classes=2,n_clusters_per_class=1,random_state=41,hypercube=False,class_sep=15)
```

```python
import matplotlib.pyplot as plt

plt.scatter(x[:,0],x[:,1],c=y,cmap='winter',s=50)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
lr = LogisticRegression()
lr.fit(x,y)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
m1 =  -(lr.coef_[0][0]/lr.coef_[0][1])
b1 = -(lr.intercept_/lr.coef_[0][1])
```

```python
x_input = np.linspace(-3,3,500)
y_input = m1 * x_input + b1
```

```python

def gd(x,y):

    x=np.insert(x,0,1,axis=1)
    weights = np.ones(x.shape[1])
    lr = 0.5
    for i  in range(5000):
        y_hat=sigmoid(np.dot(x,weights))
        weights = weights + lr*(np.dot((y-y_hat),x)/x.shape[0])

    return weights[1:],weights[0]

 
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
coef_,intercept_ = gd(x,y)
```

```python
m =-(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input1 = np.linspace(-3,3,500)
y_input1 = m*x_input1 + b
```
